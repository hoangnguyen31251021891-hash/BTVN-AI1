```python
#2.11
import numpy as np
import skfuzzy as fuzzy
from skfuzzy import control as ctrl
duongdi = ctrl.Antecedent(np.arange(0, 21, 1), 'duongdi')
khachdat = ctrl.Antecedent(np.arange(0, 101, 1), 'khachdat')
troixau = ctrl.Antecedent(np.arange(0, 11, 1), 'troixau')
giatien = ctrl.Consequent(np.arange(0, 101, 1), 'giatien')
duongdi['gan'] = fuzzy.trimf(duongdi.universe, [0, 0, 10])
duongdi['vua'] = fuzzy.trimf(duongdi.universe, [5, 10, 15])
duongdi['xa'] = fuzzy.trimf(duongdi.universe, [10, 20, 20])
khachdat['it'] = fuzzy.trimf(khachdat.universe, [0, 0, 50])
khachdat['on'] = fuzzy.trimf(khachdat.universe, [20, 50, 80])
khachdat['nhieu'] = fuzzy.trimf(khachdat.universe, [50, 100, 100])
troixau['ok'] = fuzzy.trimf(troixau.universe, [0, 0, 5])
troixau['te'] = fuzzy.trimf(troixau.universe, [3, 7, 10])
giatien['be'] = fuzzy.trimf(giatien.universe, [0, 0, 30])
giatien['vua'] = fuzzy.trimf(giatien.universe, [20, 50, 80])
giatien['cao'] = fuzzy.trimf(giatien.universe, [50, 80, 100])
giatien['khung'] = fuzzy.trimf(giatien.universe, [70, 100, 100])
rule1 = ctrl.Rule(duongdi['gan'] & khachdat['it'], giatien['be'])
rule2 = ctrl.Rule(duongdi['vua'] & khachdat['on'], giatien['vua'])
rule3 = ctrl.Rule(khachdat['nhieu'] | troixau['te'], giatien['cao'])
rule4 = ctrl.Rule(duongdi['xa'] & khachdat['nhieu'] & troixau['te'], giatien['khung'])
maytinh = ctrl.ControlSystem([rule1, rule2, rule3, rule4])
chay = ctrl.ControlSystemSimulation(maytinh)
chay.input['duongdi'] = float(input("Quang duong di: "))
chay.input['khachdat'] = float(input("Nhu cau khach: "))
chay.input['troixau'] = float(input("Thoi tiet xau: "))
chay.compute()
print(f"Gia cuoc tinh duoc: {chay.output['giatien']:.2f}")
giatien.view(sim=chay)
```
